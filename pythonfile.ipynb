```python
ax= sns.countplot(x='gender' , data= df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
#checking info of placements 
ax = sns.countplot(x = 'status', data = df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
#placement rate according to gender
ax = sns.countplot(x='gender', hue='status', data=df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
#workex vs status(impact of work experience)
ax = sns.countplot(x='workex', hue='status', data=df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
#academic percentage impacts(boxplot)
sns.boxplot(x='status', y='degree_p',data=df)
```

```python
#correlation of academic percentage heatmape(multivariate)
#impact and connection of 10th,12th and degree marks on salary
#selecting only numeric column to make heatmap
numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns
sns.heatmap(df[numerical_cols].corr(), annot=True, cmap='coolwarm')
```

```python
#12th stream vs MBA vs STATUS
sns.catplot(x='specialisation',hue='status',col='hsc_s',data= df,kind='count',height=4,aspect=1)
```

```python
#graduation degree type vs salary i.e which field is getting more package
#taking data of placed people only
placed_df = df[df['status'] == 'Placed']

ax = sns.barplot(x='degree_t', y='salary', data=placed_df, errorbar=None,hue='degree_t', palette='Set2',legend= False)
for bars in ax.containers:
    ax.bar_label(bars, fmt='%.0f')
```

```python
#etest_p(employability test percentage) distribution,chart:sns.histplot(with KDE)
sns.histplot(x='etest_p',data=df,kde=True,color='purple')
```

```python
#work experience vs salary(KDE/density plot)
sns.kdeplot(data=placed_df,x='salary',hue='workex',fill=True, common_norm=False, palette='crest',alpha=0.5)
```

```python
#salary based on gender
df.groupby(['gender'],as_index=False)['salary'].sum().sort_values(by='salary',ascending=False)
```

```python
#average salary stat of male and female
plt.figure(figsize=(6, 4))
ax = sns.barplot(x='gender', y='salary', data=df[df['status'] == 'Placed'], errorbar=None, hue='gender', palette='Set1', legend=False)

# clear labels of packages on bar
for bars in ax.containers:
    ax.bar_label(bars, fmt='%.0f')

plt.title('Average Salary Package by Gender')
plt.show()
```

```python

# 📊 Executive Summary & Final Conclusion
**Project Title:** Campus Placement Data Analysis & Insights
**Dataset Size:** 215 Students | 14 Core Features
## 🔍 Key Insights & Findings
### 1. Gender Distribution & Impact
 * The batch is highly skewed toward male students, with **Males (139)** outnumbering **Females (76)** by nearly a 2:1 ratio.
 * While the total cumulative salary is heavily driven by male students due to their higher volume, the average salary package remains competitive between both genders.
### 2. Academic Streams & Corporate Demand
 * **Highest Packages:** Students who pursued **Science & Technology (Sci&Tech)** during graduation command the highest average salary packages (**~3,14,610**).
 * **Volume Placements:** **Commerce & Management (Comm&Mgmt)** students account for the highest volume of placed candidates, pulling in a stable average package of approximately **2,78,627**.
### 3. Work Experience (The Golden Ticket)
 * The Kernel Density Estimate (KDE) plot clearly reveals that candidates with prior **Work Experience (workex = Yes)** enjoy a massive competitive edge. They show significantly higher placement conversion rates and are more likely to land in the elite, higher-bracket salary packages (outliers).
### 4. Specialisation Performance (Marketing & Finance vs. HR)
 * Students specializing in **Marketing & Finance (Mkt&Fin)** face much higher corporate demand and recruitment consistency compared to those opting for **Marketing & HR (Mkt&HR)**.
## 💡 Strategic Recommendations for Students
 1. **Prioritize Technical Skills:** To secure top-tier, high-paying corporate roles, building a strong base in tech-driven fields or analytical tools during graduation is crucial.
 2. **Gain Early Experience:** Securing short-term internships or full-time jobs prior to pursuing an MBA drastically maximizes overall placement value and negotiation power.
 3. **Select Specialisations Intentionally:** Based on current historical trends within this data, **Marketing & Finance** stands out as a safer and more rewarding career choice.
## 🛠️ Data Cleaning Actions Taken (Summary)
 * The sl_no column was dropped entirely as it functioned purely as a serial identifier and held no statistical predictive value for EDA or Machine Learning.
 * Missing values (NaN) in the salary column were strategically imputed with 0 for unplaced students (status = Not Placed) to preserve the mathematical integrity of our data distributions.
```
